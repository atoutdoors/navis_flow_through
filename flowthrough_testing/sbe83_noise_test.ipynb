```python
import pandas as pd
import numpy as np
from scipy.signal import butter, filtfilt
import matplotlib.pyplot as plt
```

```python
floatnum = 1569
# Insert location to Pickle File with timeseries data
data = pd.read_pickle("G:\\Shared drives\\WHOI Argo\\BGC\\Float Tracking\\pressure_test_data\\valve\\1659\\F1659_20240905.pkl")
```

```python
b, a = butter(4,0.1, 'highpass')

data['O2_PHASE_DELAY_filtered'] = filtfilt(b,a,data['O2_PHASE_DELAY'])
```

```python
plt.scatter(data['DATE'],data['O2_PHASE_DELAY_filtered'])
```

```python
plt.scatter(data['DATE'],data['O2_PHASE_DELAY'])
```

```python
# Take 5 Subsets from dataframe and get standard deviation in nanoseconds
std0=(data.iloc[51:61]['O2_PHASE_DELAY_filtered'].std())*1000
std1=(data.iloc[11:21]['O2_PHASE_DELAY_filtered'].std())*1000
std2=(data.iloc[31:41]['O2_PHASE_DELAY_filtered'].std())*1000
std3=(data.iloc[21:31]['O2_PHASE_DELAY_filtered'].std())*1000
std4=(data.iloc[41:51]['O2_PHASE_DELAY_filtered'].std())*1000
# Mean of the five standard deviations
stdev = np.mean([std0,std1,std2,std3,std4])
print('Standard Deviation of Noise from SBE83 on Float ' + str(floatnum) + ' is ' + str(stdev) + ' nanoseconds')
```
